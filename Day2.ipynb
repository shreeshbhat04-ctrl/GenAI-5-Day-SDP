```python
import torch
from transformers import pipeline
from huggingface_hub import login
from google.colab import userdata
import os
```

```python
try:
    generator = pipeline(
        task="text-generation",
        model="distilgpt2"
    )

    print("Model loaded successfully.")
except Exception as e:
    print(f"Error loading model: {e}")
```

```python
print("\n--- Few-Shot Prompting ---")
few_shot_prompt = (
        "Classify the sentiment of the following reviews.\n\n"
        "Review: I loved the movie, it was fantastic!\n"
        "Sentiment: positive\n\n"
        "Review: The movie was boring and too long.\n"
        "Sentiment: negative\n\n"
        "Review: The story was okay but the acting was bad.\n"
        "Sentiment:"
    )
```

```python
result = generator(
        few_shot_prompt,
        max_new_tokens=20,
        do_sample=True
    )

    # Print result
print(result[0]["generated_text"])
print("-" * 40)
```

```python
import torch
from diffusers import StableDiffusionPipeline
from PIL import Image
```

```python
model_id = "runwayml/stable-diffusion-v1-5"

print("Loading model... this may take a minute.")

try:
    pipe = StableDiffusionPipeline.from_pretrained(
        model_id,
        torch_dtype=torch.float16
    )

    # Move the model to the GPU for fast generation
    pipe = pipe.to("cuda")

    print("✅ Model loaded successfully and moved to GPU.")

except Exception as e:
    print(f"❌ Error loading model: {e}")
    print("Ensure your Runtime is set to GPU (Runtime > Change runtime type > T4 GPU)")
```

```python
def generate_image(prompt):
    print(f"\nGenerating image for: '{prompt}'...")

    # Inference: The model creates the image here
    image = pipe(prompt).images[0]

    return image
```

```python
# Define your text prompt
text_prompt = "A futuristic cyberpunk city with neon lights, cinematic lighting, highly detailed"

# Run generation
try:
    generated_image = generate_image(text_prompt)

    # Display the image (for Colab/Jupyter)
    display(generated_image)

    # Save the image locally
    generated_image.save("generated_image.png")
    print("\nImage saved as 'generated_image.png'")

except NameError:
    print("Pipeline not loaded. Check GPU settings.")
except Exception as e:
    print(f"Error during generation: {e}")
```
